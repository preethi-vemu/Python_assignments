```python
# Question 9: Count frequency of each character in string 'hello'
# solution :

text="hello"
frequency={}
for char in text:
    if char in frequency:
        frequency[char]=frequency[char]+1
    else:
        frequency[char]=1
print("Frequency:",frequency)
```

```python
#Question 33: Create a dictionary with complex numbers: {'z1': 3+4j, 'z2': 1+2j}
# solution :

data={"z1":3+4j,"z2":1+2j}
print("Dictionary:",data)
```

```python
# Question 34: Find magnitude of each complex number
# solution :

for key in data:
    number=data[key]
    # Magnitude = square root of (real part squared + imaginary part squared)
    magnitude=(number.real**2+number.imag**2)**0.5
    print(key,"magnitude:",magnitude)
```

```python
# Question 43: Create a dictionary with nested lists: {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}
# solution :

data={"matrix":[[1,2],[3,4]],"vector":[5,6,7]}
print("Dictionary:",data)
```

```python
# Question 44: Find sum of each nested list
# solution :

data={"matrix":[[1,2],[3,4]],"vector":[5,6,7]}
for row in data["matrix"]:
    total=0
    for number in row:
        total+=number
    print("Matrix row sum:",total)
total=0
for number in data["vector"]:
    total+=number
print("Vector sum:",total)
```

```python
# Question 49: Create a dictionary with nested sets: {'groups': {{1, 2, 3}, {4, 5, 6}}, 'categories': {{'a', 'b'}, {'c', 'd'}}}
# solution :

data={"groups":[{1,2,3},{4,5,6}],"categories":[{"a","b"},{"c","d"}]}
print("Dictionary:",data)
```

```python
# Question 50: Find union of all nested sets
# solution :

data={"groups":[{1,2,3},{4,5,6}],"categories":[{"a","b"},{"c","d"}]}
groups_union=data["groups"][0].union(data["groups"][1])
categories_union=data["categories"][0].union(data["categories"][1])
print("Groups union:",groups_union)
print("Categories union:",categories_union)
```
