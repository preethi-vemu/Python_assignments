The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:

Problem 7.2: Caesar Cipher (Hard)

Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3

     . 'A' becomes 'D'
     . 'Hello' becomes 'Khoor'

Your program should:

1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

Hint: Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

```python
message = input("Enter the message: ")
shift = int(input("Enter the shift value: "))

choice = input("Enter E to Encrypt or D to Decrypt: ")

result = ""

if choice.upper() == "D":
    shift = -shift

for char in message:
    if char.isalpha():
        if char.isupper():
            start = ord('A')
        else:
            start = ord('a')

        new_char = chr((ord(char) - start + shift) % 26 + start)
        result += new_char
    else:
        result += char

if choice.upper() == "E":
    print("Encrypted message:", result)
elif choice.upper() == "D":
    print("Decrypted message:", result)
else:
    print("Invalid choice!")
```

Hint: Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

```python
balance = 10000
pin = "1234"
attempts = 3
history = []

# PIN verification
while attempts > 0:
    entered_pin = input("Enter your PIN: ")

    if entered_pin == pin:
        print("\nLogin successful!")
        break
    else:
        attempts -= 1
        print("Incorrect PIN. Attempts left:", attempts)

if attempts == 0:
    print("Your account is locked. Please try again later.")

else:
    while True:
        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Change PIN")
        print("5. Transaction History")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            print("Current balance:", balance)

        elif choice == "2":
            try:
                amount = float(input("Enter withdrawal amount: "))

                if amount <= 0:
                    print("Amount must be greater than zero.")
                elif amount > balance:
                    print("Insufficient balance.")
                else:
                    balance -= amount
                    print("Withdrawal successful.")
                    print("Remaining balance:", balance)
                    history.append("Withdrawn: " + str(amount))

            except ValueError:
                print("Invalid amount. Enter a number.")

        elif choice == "3":
            try:
                amount = float(input("Enter deposit amount: "))

                if amount <= 0:
                    print("Amount must be greater than zero.")
                else:
                    balance += amount
                    print("Deposit successful.")
                    print("Current balance:", balance)
                    history.append("Deposited: " + str(amount))

            except ValueError:
                print("Invalid amount. Enter a number.")

        elif choice == "4":
            old_pin = input("Enter your current PIN: ")

            if old_pin == pin:
                new_pin = input("Enter your new 4-digit PIN: ")

                if len(new_pin) == 4 and new_pin.isdigit():
                    if new_pin != pin:
                        pin = new_pin
                        print("PIN changed successfully.")
                        history.append("PIN changed")
                    else:
                        print("New PIN must differ from old PIN.")
                else:
                    print("PIN must contain exactly 4 digits.")
            else:
                print("Incorrect current PIN.")

        elif choice == "5":
            print("\n===== LAST 5 TRANSACTIONS =====")

            if len(history) == 0:
                print("No transactions yet.")
            else:
                for transaction in history[-5:]:
                    print(transaction)

        elif choice == "6":
            print("Thank you for using the ATM!")
            break

        else:
            print("Invalid choice. Please select 1 to 6.")

        # Keep only the last 5 transactions
        history = history[-5:]
```

Problem 11.3: Spiral Matrix (Hard)

Print numbers in a spiral pattern for n×n matrix.

Hint: Use four boundaries (top, bottom, left, right). Fill in order: left to right (top row), top to bottom (right col), right to left (bottom row), bottom to top (left col). Shrink boundaries.


```python
n = int(input("Enter n: "))

matrix = [[0] * n for _ in range(n)]

top, bottom = 0, n - 1
left, right = 0, n - 1
num = 1

while top <= bottom and left <= right:

    for i in range(left, right + 1):
        matrix[top][i] = num
        num += 1
    top += 1

    for i in range(top, bottom + 1):
        matrix[i][right] = num
        num += 1
    right -= 1

    if top <= bottom:
        for i in range(right, left - 1, -1):
            matrix[bottom][i] = num
            num += 1
        bottom -= 1

    if left <= right:
        for i in range(bottom, top - 1, -1):
            matrix[i][left] = num
            num += 1
        left += 1

for row in matrix:
    print(*row)
```

Hint: Use 3x3 list. Loop for turns. Display board, get position, check validity, place X or O. Check win conditions (rows, columns, diagonals) after each move.


```python
board = [" "] * 9
wins = [(0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)]


def show_board():
    print()
    for r in range(3):
        print(" " + " | ".join(board[r * 3:r * 3 + 3]))
        if r < 2:
            print("---+---+---")
    print()


def check_winner(player):
    for a, b, c in wins:
        if board[a] == board[b] == board[c] == player:
            return True
    return False


print("Tic-Tac-Toe! Positions are numbered 1-9:")
print(" 1 | 2 | 3")
print("---+---+---")
print(" 4 | 5 | 6")
print("---+---+---")
print(" 7 | 8 | 9")

player = "X"
moves = 0
while moves < 9:
    pos = input(f"Player {player}, choose position (1-9): ")

    if not pos.isdigit() or not 1 <= int(pos) <= 9:
        print("Invalid input! Choose a number between 1 and 9.")
        continue

    pos = int(pos) - 1
    if board[pos] != " ":
        print("That position is already taken!")
        continue

    board[pos] = player
    moves += 1
    show_board()

    if check_winner(player):
        print(f"Player {player} wins!")
        break

    player = "O" if player == "X" else "X"
else:
    print("It's a draw!")
```

Implement the Sieve of Eratosthenes algorithm to find all primes up to n.

Hint: Create boolean list (all True). For each number i from 2 to √n, mark all multiples of i as False (not prime). Print numbers that remain True.

```python
n = int(input("Enter n: "))

is_prime = [True] * (n + 1)
is_prime[0] = False
if n >= 1:
    is_prime[1] = False

for i in range(2, int(n ** 0.5) + 1):
    if is_prime[i]:
        for j in range(i * i, n + 1, i):
            is_prime[j] = False

print(f"Prime numbers up to {n}:")
for i in range(2, n + 1):
    if is_prime[i]:
        print(i, end=" ")
print()
```

Problem 11.9: Anagram Checker (Medium)

Check if two strings are anagrams of each other.

Hint: Sort both strings and compare, or count frequency of each character in both strings and compare.

```python
s1 = input("Enter first string: ").replace(" ", "").lower()
s2 = input("Enter second string: ").replace(" ", "").lower()

if sorted(s1) == sorted(s2):
    print("The strings are anagrams")
else:
    print("The strings are not anagrams")
```

Problem 11.10: Sudoku Validator (Hard)

Validate if a given 9×9 Sudoku solution is correct.

Hint: Check three conditions: (1) Each row has unique 1-9, (2) Each column has unique 1-9, (3) Each 3x3 box has unique 1-9. Use sets to check uniqueness.



```python

def is_valid_sudoku(grid):
    full_set = set(range(1, 10))

    # (1) Check each row
    for row in grid:
        if set(row) != full_set:
            return False

    # (2) Check each column
    for col in range(9):
        if set(grid[row][col] for row in range(9)) != full_set:
            return False

    # (3) Check each 3x3 box
    for box_row in range(0, 9, 3):
        for box_col in range(0, 9, 3):
            box = set()
            for i in range(3):
                for j in range(3):
                    box.add(grid[box_row + i][box_col + j])
            if box != full_set:
                return False

    return True


sudoku = [
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9],
]

print("Grid 1:", "Valid Sudoku" if is_valid_sudoku(sudoku) else "Invalid Sudoku")

# Change one cell to create a duplicate and test again
sudoku[0][0] = 3
print("Grid 2:", "Valid Sudoku" if is_valid_sudoku(sudoku) else "Invalid Sudoku")
```
