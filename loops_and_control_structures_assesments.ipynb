Problem 3.5: Armstrong Number (Hard)

Write a program to check if a number is an Armstrong number (also called Narcissistic number).

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

Examples:

153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 (Armstrong number)

9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)

Hint: First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

```python
num = int(input("Enter a number: "))

power = len(str(num))
total_sum = 0
temp = num
while temp > 0:
    digit = temp % 10              
    total_sum += digit ** power     
    temp = temp // 10              
if total_sum == num:
    print("armstrong")
else:
    print("Not a armstrong")
```

Problem 4.6: Diamond Pattern (Hard)

Print a diamond pattern for n rows (n should be odd):

        *  
       ***
      *****
     *******
    *********
     *******
      *****
       ***
        *

(Example shown for n=9)

Hint: Print pyramid (upper half) for rows 1 to (n+1)//2, then print inverted pyramid (lower half) for remaining rows.

```python
n = int(input("enter number of rows: "))  # 5
# Then I am calculating the middle row using k.
k = (n + 1) // 2
if n % 2 == 1:
    for i in range(k - 1, -k, -1):  # 2 1 0 -1 -2
        
        # The first inner loop prints spaces before the stars.
        for j in range(abs(i)):
            print(" ", end=" ")

        # The second inner loop prints stars in each row.
        # The number of stars increases until the middle row and then decreases.
        for star in range(2 * (k - abs(i)) - 1):
            print("*", end=" ")
        print()
else:
    print("Enter the odd number")
```

Problem 6.2: GCD Calculator (Medium)

Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

Algorithm:

1. Divide the larger number by the smaller number
2. Replace the larger number with the smaller number
3. Replace the smaller number with the remainder
4. Repeat until the remainder is 0
5. The GCD is the last non-zero remainder

Hint: While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
num1=a
num2=b
while b != 0:
    a, b = b, a % b
print(f"The GCD of {num1} and {num2} is: {a}")
```

Problem 6.3: Perfect Number Checker (Hard)

Write a program to check if a number is a perfect number.

A perfect number is a positive integer that is equal to the sum of its proper positive divisors (excluding the number itself).

Example: 6 is perfect because 6 = 1 + 2 + 3

Example: 28 is perfect because 28 = 1 + 2 + 4 + 7 + 14

Hint: Loop from 1 to n-1. If i divides n (n%i==0), add i to sum. Compare sum with n.

```python
num = int(input("Enter a number: "))

if num <= 0:
    print(num, "is not a perfect number.")
else:
    divisor_sum = 0

    for i in range(1, num):
        if num % i == 0:
            divisor_sum += i

    if divisor_sum == num:
        print(num, "is a perfect number.")
    else:
        print(num, "is not a perfect number.")
```

Problem 6.5: Number to Words (Hard)

Write a program that converts a number (0-999) to words.

Examples:

0: Zero,
25: Twenty Five,
123: One Hundred Twenty Three,
500: Five Hundred.

Hint: Create lists for ones, teens, tens, etc. Extract hundreds, tens, ones digits. Convert each part to words and combine.

```python
num = int(input("Enter a number (0-999): "))

# Word lookup lists (using empty strings for index 0 to align indices with numbers)
ones = ["", "One", "Two", "Three", "Four", "Five", "Six", "Seven", "Eight", "Nine"]
teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen", "Fifteen", 
         "Sixteen", "Seventeen", "Eighteen", "Nineteen"]
tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty", 
        "Sixty", "Seventy", "Eighty", "Ninety"]

if num == 0:
    print("Zero")
elif num < 0 or num > 999:
    print("Please enter a number between 0 and 999.")
else:
    words = []

    hundreds = num // 100
    remainder = num % 100

    if hundreds > 0:
        words.append(ones[hundreds] + " Hundred")

    if 10 <= remainder <= 19:
        words.append(teens[remainder - 10])
    else:
        tens_digit = remainder // 10
        ones_digit = remainder % 10
        
        if tens_digit > 0:
            words.append(tens[tens_digit])
        if ones_digit > 0:
            words.append(ones[ones_digit])

    print(" ".join(words))
```

Problem 6.6: Strong Number Checker (Hard)

Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

Hint: Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

```python
num = int(input("Enter a number: "))

original_num = num
total_sum = 0
temp = num

while temp > 0:
    digit = temp % 10 
    
    fact = 1
    for i in range(1, digit + 1):
        fact *= i
        
    total_sum += fact 
    temp = temp // 10 
if total_sum == original_num:
    print(original_num, "is a Strong number.")
else:
    print(original_num, "is not a Strong number.")
```
