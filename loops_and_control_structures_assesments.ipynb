Problem 1.2: Leap Year Checker (Medium)

Write a program to check if a given year is a leap year or not.

Rules:

A year is a leap year if it is divisible by 4

However, if it is divisible by 100, it is NOT a leap year

Exception: If it is also divisible by 400, then it IS a leap year

Examples:

2000: Leap year (divisible by 400)

1900: Not a leap year (divisible by 100 but not 400)

2024: Leap year (divisible by 4 but not 100)

Hint: Check conditions in order: first divisible by 400, then by 100, then by 4. Use modulo operator (%) to check divisibility.

```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

Problem 3.1: Reverse a Number (Medium)

Write a program that reverses the digits of a given number using a while loop.

Example: Input: 12345, Output: 54321

Hint: Initialize reverse=0. While n>0: get last digit (n%10), add to reverse (reverse=reverse*10+digit), remove last digit (n=n//10)

```python
num=int(input("enter the number"))
temp=num
rev=0
while temp>0:
    digit=temp%10
    rev=rev*10+digit
    temp//=10
print(rev)
```

Problem 3.3: Number Guessing Game (Medium)

Write a number guessing game where:

1. The program generates a random number between 1 and 100

2. The user has to guess the number

3. After each guess, the program tells the user if their guess is too high or too low

4. The game continues until the user guesses correctly

5. Display the number of attempts taken

Hint: Use import random and random.randint(1, 100) to generate number. Use while loop with condition guess != target. Count attempts.

```python
import random

n = random.randint(1, 100)
count = 0

while count < 5:
    guess = int(input("Guess the number: "))
    count += 1

    if guess == n:
        print("Correct!")
        break
    elif guess > n:
        print("Too high")
    else:
        print("Too low")

print("Attempts:", count)
```

Problem 3.4: Palindrome Checker (Medium)

Write a program to check if a given number is a palindrome or not.

A palindrome number reads the same forwards and backwards.

Examples: 121, 12321, 1331 are palindromes

Hint: Store original number, reverse it (like Problem 3.1), then compare original with reversed number.

```python
number=int(input("Guess the number: "))
temp=number
rev=0
while temp>0:
    digit=temp%10
    rev=rev*10+digit
    temp//=10

if number==rev:
    print(f"{number} is Palindrome")
else:
    print(f"{number} is Not a Palindrome")
```

Problem 3.5: Armstrong Number (Hard)

Write a program to check if a number is an Armstrong number (also called Narcissistic number).

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

Examples:

153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 (Armstrong number)

9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)

Hint: First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

```python
num = int(input("Enter a number: "))

power = len(str(num))
total_sum = 0
temp = num
while temp > 0:
    digit = temp % 10              
    total_sum += digit ** power     
    temp = temp // 10              
if total_sum == num:
    print("armstrong")
else:
    print("Not a armstrong")
```

Problem 4.6: Diamond Pattern (Hard)

Print a diamond pattern for n rows (n should be odd):

        *  
       ***
      *****
     *******
    *********
     *******
      *****
       ***
        *

(Example shown for n=9)

Hint: Print pyramid (upper half) for rows 1 to (n+1)//2, then print inverted pyramid (lower half) for remaining rows.

```python
n = int(input("enter number of rows: "))  # 5
# Then I am calculating the middle row using k.
k = (n + 1) // 2
if n % 2 == 1:
    for i in range(k - 1, -k, -1):  # 2 1 0 -1 -2
        
        # The first inner loop prints spaces before the stars.
        for j in range(abs(i)):
            print(" ", end=" ")

        # The second inner loop prints stars in each row.
        # The number of stars increases until the middle row and then decreases.
        for star in range(2 * (k - abs(i)) - 1):
            print("*", end=" ")
        print()
else:
    print("Enter the odd number")
```
